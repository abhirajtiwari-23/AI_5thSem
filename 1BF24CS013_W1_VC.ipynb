```python
# Simple Reflex Vacuum Cleaner Agent

def simple_reflex_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    else:
        return "Move Left"


# Input
location = input("Enter location (A/B): ")
status = input("Enter room status (Clean/Dirty): ")

# Agent action
action = simple_reflex_agent(location, status)

# Output
print("Current Location:", location)
print("Room Status:", status)
print("Action:", action)
```

```python
# Goal-Based Vacuum Cleaner Agent

rooms = {
    "A": input("Enter status of Room A (Clean/Dirty): "),
    "B": input("Enter status of Room B (Clean/Dirty): ")
}

location = input("Enter starting location (A/B): ")

print("\nInitial State:")
print("Room A:", rooms["A"])
print("Room B:", rooms["B"])
print("Vacuum Location:", location)

# Clean Room A
if rooms["A"] == "Dirty":
    if location != "A":
        print("Move to Room A")
        location = "A"

    print("Suck dirt from Room A")
    rooms["A"] = "Clean"

# Clean Room B
if rooms["B"] == "Dirty":
    if location != "B":
        print("Move to Room B")
        location = "B"

    print("Suck dirt from Room B")
    rooms["B"] = "Clean"

# Goal test
if rooms["A"] == "Clean" and rooms["B"] == "Clean":
    print("\nGoal Achieved!")
    print("Both rooms are clean.")
```
